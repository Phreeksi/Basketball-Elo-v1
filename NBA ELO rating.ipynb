```python
def season_data(key, year = [2022]):
    ''''
    Allows you to pull season data from ballislife, need to pass your key and the years you want.
    Each year takes about 7 minutes but you can speed things up if you have a better plan
    Default year is 2022
    '''
    api = BalldontlieAPI(api_key= key)
    cur = None
    all_data = []
    while True:
        new_games = api.nba.games.list(seasons= year, per_page=100, cursor=cur)
        new_row = [ng.model_dump() for ng in new_games.data]
        all_data.extend(new_row)
        cur = new_games.meta.next_cursor
        if cur is None:
            break
        time.sleep(31) 
    df = pd.json_normalize(all_data,sep='.')
    return df
```

```python
data = df[[	'season','date','away','home','score_away','score_home','moneyline_away','moneyline_home']]
data['home_win'] = (data['score_home'] > data['score_away']).astype(int)
data['date'] = pd.to_datetime(data.date)
data['away'] = [x.upper() for x in data.away]
data['home'] = [x.upper() for x in data.home]
data['score_home'] = data['score_home'].astype(int)
data['score_away'] = data['score_away'].astype(int)
data2022 = data[data['season'] == 2022]
```

```python
# Define starting Elo for all teams, Home field advantage, and K. 
all_elo = {}
r0 = 1500
K = 20
H = 60
for x in data.home.unique():
    all_elo[x] = r0
all_elo
```

```python
def win_prob(home_rating, away_rating):
    ''''
    Probability of Winning for the home team when you provide the ratings and homefield advantage
    '''
    p_home = 1 / (1 + 10 ** -((home_rating + H - away_rating)/400))
    return p_home
```

```python
#testing win_prob on elo
print(win_prob(1500, 1500))
print(win_prob(1600, 1500))    
print(win_prob(1400, 1600))  
```

```python
def update_rating(home, away, home_win):
    ''''
    Update Rating of the Winner
    '''
    #gets the ratings
    home_rating = all_elo[home]
    away_rating = all_elo[away]
    #gets the probabilty of home team winning
    p_home = win_prob(home_rating, away_rating)
    #updates the new rating
    home_rating += K * (home_win - p_home)
    away_rating += K * ((1 - home_win) - (1 - p_home))
    #makes the new rating permenate
    all_elo[home] = home_rating
    all_elo[away] = away_rating
```

```python
update_rating("LAL", "BOS", 1)
print(all_elo["LAL"], all_elo["BOS"])
```

```python
#save predictions
pred = []
for g in data2022.itertuples():
    # our predictions
    home_rat = all_elo[g.home]
    away_rat = all_elo[g.away]
    home_pred = win_prob(home_rat, away_rat)
    #saving our predictions
    pred.append(home_pred)
    #updating rating
    update_rating(g.home, g.away, g.home_win)

pred
```

```python
data2022["h_prob"] = pred
data2022["a_prob"] = 1 - data2022["h_prob"]
```

```python
def brier_baseline(data):
    baseline = data.home_win.mean()
    brier = np.mean((baseline - data.home_win) ** 2)
    return brier
```

```python
def brier_model(data):
    prediction = data.h_prob
    brier_score = np.mean((prediction - data.home_win) ** 2)
    return brier_score
```

```python
def price_to_prob(price):
    
    if price > 0:
        p_hat = 100 / (price + 100)
    else:
        p_hat = -price / (-price + 100)
    return p_hat
```

```python
def implied_prob(ph_home, ph_away):
    s = ph_home + ph_away
    p_home = ph_home / s
    p_away = ph_away / s
    return [p_home, p_away]
```

```python
def american_to_decimal(price):
    if price > 0:
        sigma = 1 + (price / 100) 
    else:
        sigma = 1 + (100/ - price)
    return sigma
```

```python
def expected_value(prob_model, price):
    sigma = american_to_decimal(price)
    beta = sigma - 1
    return prob_model * beta - (1 - prob_model)
```

```python
def Kelly_fraction(prob_model, price):
    sigma = american_to_decimal(price)
    beta = sigma - 1
    f = expected_value(prob_model, price) / beta
    return min(max(f,0),0.05)
```

```python
def cash_multi(odds):
    if odds > 0:
        return (odds / 100)
    else:
        return (100/ abs(odds))
```

```python
def simulation(h_kelly, a_kelly, h_win, h_odds, a_odds, cash = 100):
    if h_kelly == a_kelly:
        return 0
    elif h_kelly > a_kelly:
        if h_win:
            return h_kelly * cash * cash_multi(h_odds)
        else:
            return -h_kelly * cash
    else:
        if h_win:
            return -a_kelly * cash 
        else:
            return a_kelly * cash * cash_multi(a_odds)
```

```python
money_made = []
for h_kelly, a_kelly, h_win, h_odds, a_odds in data2022[['home_kelly','away_kelly', 'home_win', 'moneyline_home', 'moneyline_away']].iloc:
    money_made.append(simulation(h_kelly, a_kelly, h_win, h_odds, a_odds))
```

```python
data2022['money_made'] = money_made
```
